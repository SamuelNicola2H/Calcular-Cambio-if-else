## Las variables
1. CP --> Costo de Producto
2. CD --> Cantidad de dinero
3. Cambio

## Formulas
$Cambio = CD - CP$

## Diagrama de fliujo Lucidchart
<img src="https://raw.githubusercontent.com/SamuelNicola2H/Calcular-Cambio-if-else/refs/heads/main/DF%20if%20else%20cambio2.png" width="400" height="500" alt="Descripción de la imagen">

```python
# para cuando es verdadero (Si)
# Calculando el cambio usando if else
# datos de entrada CD, CP
# Mostrar Cambio si es falso
# Si es verdadero mostrar no alcanza
CP = float(input("Ingrese el costo del producto: "))
CD = float(input("Ingrese la cantidad de dinero: "))
if CD < CP:
    print("No alcanza")
else:
    Cambio = CD - CP
    print("El cambio es: ", Cambio)
print("Fin del programa")
print("Programa realizado por Samuel Nicola NL = 40")
```

```python
# para cuando es falso (No)
# Calculando el cambio usando if else
# datos de entrada CD, CP
# Mostrar Cambio si es falso
# Si es verdadero mostrar no alcanza
CP = float(input("Ingrese el costo del producto: "))
CD = float(input("Ingrese la cantidad de dinero: "))
if CD < CP:
    print("No alcanza")
else:
    Cambio = CD - CP
    print("El cambio es: ", Cambio)
print("Fin del programa")
print("Programa realizado por Samuel Nicola NL = 40")
```
